```python
import os
def generate_verilog_adder(n):
    assert n >= 1, "Bit-width must be at least 1"

    module_name = f"{n}_bit_adder"
    inputs = ", ".join([f"a{i}" for i in range(n)] + [f"b{i}" for i in range(n)] + ["cin"])
    outputs = ", ".join([f"sum{i}" for i in range(n)] + ["cout"])

    # Module header
    verilog = f"// File: {module_name}.v\n"
    verilog += f"module {module_name} (\n    input {inputs};\n    output {outputs};\n);\n\n"

    # Wire declarations for carry bits
    if n > 1:
        carry_wires = ", ".join([f"c{i}" for i in range(1, n)])
        verilog += f"wire {carry_wires};\n\n"

    # Generate bitwise full adder logic
    for i in range(n):
        a, b, s = f"a{i}", f"b{i}", f"sum{i}"
        ci = "cin" if i == 0 else f"c{i}"
        co = "cout" if i == n - 1 else f"c{i+1}"

        verilog += f"// Bit {i}\n"
        verilog += f"assign {s} = {a} ^ {b} ^ {ci};\n"
        verilog += f"assign {co} = ({a} & {b}) | ({b} & {ci}) | ({a} & {ci});\n\n"

    # End module
    verilog += "endmodule\n"
    return verilog


# Create and save Verilog files for 20 to 32-bit adders
directory = "adders/"
os.makedirs(directory, exist_ok=True)

for i in range(20, 32+1):
    filename = os.path.join(directory, f"{i}-bitAdder.v")
    with open(filename, "w") as f:
        f.write(generate_verilog_adder(i))
```
